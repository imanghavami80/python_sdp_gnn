## 0) Dataset Description (Before Preprocessing)

This section provides a quick profile of the raw dataset: samples, columns, dtypes, missing values, label distribution, and descriptive statistics.

```python
from pathlib import Path
import sys
import pandas as pd
```

```python
# Resolve repo root whether notebook is run from repo root or notebooks/
REPO_ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
CSV_PATH = REPO_ROOT / 'projects/log4j/log4j-1.0.csv'
SOURCE_ROOT = REPO_ROOT / 'projects/log4j/logging-log4j1-v_1_0/src/java'
OUTPUT_DIR = REPO_ROOT / 'outputs/log4j'

CSV_PATH
```

```python
df_raw = pd.read_csv(CSV_PATH)

print('Number of samples (rows):', df_raw.shape[0])
print('Number of columns:', df_raw.shape[1])
print()
print('Columns:')
print(df_raw.columns.tolist())
print()
print('Data types:')
print(df_raw.dtypes)
print()
print('Missing values per column:')
print(df_raw.isna().sum())
print('Total missing values:', int(df_raw.isna().sum().sum()))
print()
if 'bug' in df_raw.columns:
    print('Bug label distribution:')
    print(df_raw['bug'].value_counts(dropna=False).sort_index())

df_raw.head()
```

```python
# Numeric summary statistics
df_raw.describe(include='all').T
```

## 1) Import Script Function

## 2) Run Preprocessing from Script

```python
SCALER = 'standard'  # or 'minmax'
preprocess_log4j(
    csv_path=CSV_PATH,
    source_root=SOURCE_ROOT,
    output_dir=OUTPUT_DIR,
    scaler_name=SCALER,
)
```

## 3) Inspect Outputs

```python
preprocessed_path = OUTPUT_DIR / f'log4j_preprocessed_{SCALER}.csv'
mapping_path = OUTPUT_DIR / 'log4j_name_to_source_mapping.csv'
summary_path = OUTPUT_DIR / 'log4j_preprocess_summary.txt'

df = pd.read_csv(preprocessed_path)
mapping_df = pd.read_csv(mapping_path)

print('Preprocessed shape:', df.shape)
print('Mapped classes:', int(mapping_df['source_path'].notna().sum()), '/', len(mapping_df))
print()
print('Summary:')
print(summary_path.read_text())
df.head()
```
